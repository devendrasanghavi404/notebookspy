```python

# 3. Even Number Filter
# Create a list of 10 numbers and display only the even ones.
numbers = []
for i in range(10):
    num = int(input(f'Enter number {i+1}: '))
    numbers.append(num)
print('Even numbers:', end=' ')
for num in numbers:
    if num % 2 == 0:
        print(num, end=' ')
```

```python
# 6. Maximum Finder
# Find the largest number in a list.
numbers = [2, 7, 10, 13, 18, 23, 26, 31, 34, 39]
max_num = numbers[0]
for num in numbers:
    if num > max_num:
        max_num = num
print(f"Largest: {max_num}")

```

```python
# 7. Duplicate Remover
# Remove duplicates from a list without built-in functions.
original = [1, 2, 2, 3, 4, 4, 5]
unique = []
for item in original:
    if item not in unique:
        unique.append(item)
print(unique)
```

```python
# 11. Longest String Finder
# Identify the longest string in a list.
strings = ['hello', 'world', 'how', 'are', 'you']
longest = ''
for s in strings:
    if len(s) > len(longest):
        longest = s
print(f"Longest: {longest}")
```

```python

# 12. Manual List Reversal
# Reverse a list without using reverse()
original = [int(input(f'Enter number {i+1}: ')) for i in range(5)]
reversed_list = []
for i in range(len(original)):
    reversed_list.append(original[-(i+1)])
print(f"Original: {original}")
print(f"Reversed: {reversed_list}")
```

```python

# 14. Index-Based Summation
# Sum elements at even and odd indices separately.
numbers = [int(input(f'Enter number {i+1}: ')) for i in range(5)]
even_idx_sum = 0
odd_idx_sum = 0
for i in range(len(numbers)):
    if i % 2 == 0:
        even_idx_sum += numbers[i]
    else:
        odd_idx_sum += numbers[i]
print(f"Even index sum: {even_idx_sum}")
print(f"Odd index sum: {odd_idx_sum}")

```

```python

# 15. Grade Distribution Analyzer
# Categorize 10 student averages into grade ranges.
marks = [int(input(f'Enter marks {i+1}: ')) for i in range(10)]
ranges = [(1, 30), (31, 50), (51, 70), (71, 85), (86, 100)]
print("\nAverage Marks\t\tNumber of Students")
for low, high in ranges:
    count = sum(1 for m in marks if low <= m <= high)
    print(f"{low}-{high}\t\t\t{count}")
```

```python

# 16. Digit Reverser
# Reverse each number in a list.
numbers = [int(input(f'Enter number {i+1}: ')) for i in range(5)]
reversed_nums = []
for num in numbers:
    rev = 0
    temp = num
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    reversed_nums.append(rev)
print(f"Original: {numbers}")
print(f"Reversed: {reversed_nums}")
```

```python

# 18. Element-wise Adder
# Add corresponding elements from two lists.
list1 = [int(input(f'List1 number {i+1}: ')) for i in range(5)]
list2 = [int(input(f'List2 number {i+1}: ')) for i in range(5)]
result = [list1[i] + list2[i] for i in range(5)]
print(f"Sum list: {result}")
```

```python

# 19. Odd Square Replacer
# Replace odd numbers with their squares.
numbers = [int(input(f'Enter number {i+1}: ')) for i in range(5)]
for i in range(len(numbers)):
    if numbers[i] % 2 != 0:
        numbers[i] **= 2
print(f"Modified: {numbers}")
```

```python
# 21. Vowel Remover
# Remove all vowels from each string in a list.
strings = [input(f'Enter string {i+1}: ') for i in range(5)]
no_vowels = []
for s in strings:
    new_str = ''.join(char for char in s if char.lower() not in 'aeiou')
    no_vowels.append(new_str)
print(f"Without vowels: {no_vowels}")

```

```python

# 25. Average Length Filter
# Remove strings shorter than average length.
strings = ['apple', 'banana', 'cherry', 'date', 'elderberry']
avg_len = sum(len(s) for s in strings) / len(strings)
filtered = [s for s in strings if len(s) >= avg_len]
print(f"Filtered: {filtered}")
```

```python

# 26. First-Last Vowel Remover
# Remove vowels from first and last strings only.
strings = ['apple', 'banana', 'cherry', 'date', 'elderberry']
result = []
for i, s in enumerate(strings):
    if i == 0 or i == len(strings) - 1:
        new_str = ''.join(char for char in s if char.lower() not in 'aeiou')
        result.append(new_str)
    else:
        result.append(s)
print(f"Result: {result}")

```

```python

# 30. Student Management System
# Menu-driven program for managing student records.
students = []

while True:
    print("\n=== STUDENT MANAGEMENT SYSTEM ===")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search by Name")
    print("4. Search by Roll Number")
    print("5. Update Student Details")
    print("6. Delete Student")
    print("7. Exit")

    choice = int(input("Enter choice (1-7): "))

    if choice == 1:
        name = input("Enter name: ")
        roll = input("Enter roll number: ")
        marks = input("Enter marks: ")
        students.append([name, roll, marks])
        print("✓ Student added successfully!")

    elif choice == 2:
        if not students:
            print("No students found!")
        else:
            print("\nName\t\tRoll No\t\tMarks")
            for s in students:
                print(f"{s[0]}\t\t{s[1]}\t\t{s[2]}")

    elif choice == 3:
        name = input("Enter name to search: ")
        found = [s for s in students if s[0].lower() == name.lower()]
        if found:
            for s in found:
                print(f"Name: {s[0]}, Roll: {s[1]}, Marks: {s[2]}")
        else:
            print("Student not found!")

    elif choice == 4:
        roll = input("Enter roll number: ")
        found = [s for s in students if s[1] == roll]
        if found:
            print(f"Name: {found[0][0]}, Roll: {found[0][1]}, Marks: {found[0][2]}")
        else:
            print("Student not found!")

    elif choice == 5:
        roll = input("Enter roll number to update: ")
        for s in students:
            if s[1] == roll:
                print("1. Update Name\n2. Update Roll\n3. Update Marks")
                sub = int(input("Choice: "))
                if sub == 1:
                    s[0] = input("New name: ")
                elif sub == 2:
                    s[1] = input("New roll: ")
                elif sub == 3:
                    s[2] = input("New marks: ")
                print("✓ Updated!")
                break
        else:
            print("Student not found!")

    elif choice == 6:
        roll = input("Enter roll number to delete: ")
        for s in students:
            if s[1] == roll:
                students.remove(s)
                print("✓ Student deleted!")
                break
        else:
            print("Student not found!")

    elif choice == 7:
        print("Thank you for using Student Management System!")
        break

    else:
        print("Invalid choice!")
```
