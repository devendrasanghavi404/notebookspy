```python
# 1. Write a python program to remove duplicate from a list using set.

lst = [1, 2, 3, 2, 4, 3, 5, 1]
print(f'Original List: {lst}')
lst = list(set(lst))
print(f'New List: {lst}')
```

```python
# 2. Write a program to carry out the following operations on the given set
# s = {10, 2, -3, 4, 5, 88}
# 1. number of items in set s
# 2. maximum element in set s
# 3. minimum element in set s
# 4. sum of all elements in set s
# 5. obtain a new sorted set from s, set s remaining unchanged
# 6. report whether 100 is an element of s
# 7. report whether -3 is an element of s

s = {10, 2, -3, 4, 5, 88}
print(len(s))
print(max(s))
print(min(s))
print(sum(s))
print(set(sorted(s)))
print(100 in s)
print(-3 in s)
```

```python
# 3. A Set contains names which begins either with A or with B, write a program to separate out the names into two sets,
# one containing names beginning with A and another containing names beginning with B.

s = {'alice', 'bob', 'anna', 'ben', 'alex'}
set_a = set()
set_b = set()

for name in s:
    if name.startswith('a'):
        set_a.add(name)
    elif name.startswith('b'):
        set_b.add(name)

print('Names starting with A:', set_a)
print('Names starting with B:', set_b)
```

```python
# 5. Write a program to show the difference between the two set functions discard() and remove().

s = {1, 2, 3, 4, 5}

try:
    s.remove(10)
except:
    print('remove() gave error')

# discard() does not give error
s.discard(10)
print('discard() did not give error')
```

```python
# 6. Write a program to create a set containing 10 randomly generated numbers in the range 15 to 45.
# Count how many of these numbers are less than 30. Delete all the numbers which are greater than 35.

import random
s = {random.randint(15, 45) for i in range(10)}
print('Original Set:', s)

count = 0
for num in s:
    if num < 30:
        count += 1
print('Numbers less than 30:', count)

s = {num for num in s if num <= 35}
print('After removing numbers > 35:', s)
```

```python
# 14. Take a string from user and print the unique count of vowels in the string.

st = input('Enter a string: ')
vowels = set()
for char in st.lower():
    if char in 'aeiou':
        vowels.add(char)
print('Unique vowel count:', len(vowels))
```

```python
# 17. You have a list of customers who have purchased products from your online store.
# You want to identify customers who have purchased from your store in the past month and customers who have not made any purchases in the past month.
# How can you use sets to accomplish this task?

past_month = {'john', 'mary', 'peter', 'susan'}
all_customers = {'john', 'mary', 'peter', 'susan', 'david', 'lisa'}

purchased = past_month.intersection(all_customers)
not_purchased = all_customers - past_month

print('Purchased in past month:', purchased)
print('Not purchased in past month:', not_purchased)
```

```python
# 18. You are the owner of a restaurant and you want to identify the most popular menu items among your customers.
# You have a list of all menu items that have been ordered in the past month.
# How can you use sets to identify the most popular menu items?

past_month_items = {'burger', 'pizza', 'pasta', 'salad', 'fries'}
all_items = {'burger', 'pizza', 'pasta', 'salad', 'fries', 'soup', 'steak'}

popular = past_month_items.intersection(all_items)
print('Popular items:', popular)
```

```python
# 19. You are the manager of a retail store and you want to identify which products are selling well and which products are not.
# You have a list of all products sold in the past month.
# How can you use sets to identify the top selling products and the least selling products?

past_month_products = {'shirt', 'pants', 'shoes', 'hat'}
all_products = {'shirt', 'pants', 'shoes', 'hat', 'jacket', 'socks', 'gloves'}

top_selling = {'shirt', 'pants', 'shoes'}
least_selling = all_products - past_month_products

print('Top selling:', top_selling)
print('Least selling:', least_selling)
```

```python
# 20. You are a data analyst and you want to identify the most common words used in a text document.
# You have a list of all words used in the document.
# How can you use sets to identify the most common words?

all_words = ['hello', 'world', 'hello', 'python', 'world', 'hello']
unique_words = set(all_words)

most_common = ''
max_count = 0

for word in unique_words:
    count = all_words.count(word)
    if count > max_count:
        max_count = count
        most_common = word

print('Most common word:', most_common)
print('Count:', max_count)
```

```python
# 21. You have a list of orders and their respective statuses (e.g. "pending", "shipped", "delivered"),
# and you want to count the number of orders in each status.
# How can you use sets to accomplish this task?

orders = [('Order1', 'pending'), ('Order2', 'shipped'), ('Order3', 'delivered'),
          ('Order4', 'pending'), ('Order5', 'delivered')]

pending = set([o[0] for o in orders if o[1] == 'pending'])
shipped = set([o[0] for o in orders if o[1] == 'shipped'])
delivered = set([o[0] for o in orders if o[1] == 'delivered'])

print('Pending:', len(pending))
print('Shipped:', len(shipped))
print('Delivered:', len(delivered))
```
