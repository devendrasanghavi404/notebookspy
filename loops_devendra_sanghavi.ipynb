```python
# 3. Do you know Factorial? It is the product of an integer and all the integers below it;
# e.g. factorial four (4!) is equal to 24. Take a number from user and calculate factorial of a number.

number = int(input('Enter the number: '))
position = 1
factorial = 1
while position <= number:
    factorial = factorial * position
    position += 1
print(f'Factorial: {factorial}')
```

```python
# 6. Find all prime numbers between 1 to 2000.

candidate = 2
while candidate <= 2000:
    divisor = 2
    is_prime = True
    while divisor < candidate:
        if candidate % divisor == 0:
            is_prime = False
            break
        divisor += 1
    if is_prime:
        print(candidate)
    candidate += 1

```

```python
# 7. Take a number from user and return sum of all digits of the number.

number = int(input('Enter The Number: '))
digit_sum = 0
temp = number
while temp > 0:
    remainder = temp % 10
    digit_sum = digit_sum + remainder
    temp = temp // 10
print(f'Sum of the digits: {digit_sum}')
```

```python
# 12. Write a python program to check whether a number given by user is a special number or not.
# A special number is the number if the sum of the factorial of its digit is same as the original number.

number = int(input('Enter a number: '))
original = number
factorial_sum = 0
while number > 0:
    digit = number % 10
    fact = 1
    loop_var = 1
    while loop_var <= digit:
        fact *= loop_var
        loop_var += 1
    factorial_sum += fact
    number //= 10
if original == factorial_sum:
    print(f'{original} is a special number')
else:
    print(f'{original} is not a special number')
```

```python
# 13. Write a program to print the even and odd numbers in the range from 1 to n where n is given by the user
# and also print the sum of all even numbers and odd numbers separately. (use while loop)

limit = int(input('Enter a number: '))
current = 1
even_sum = 0
odd_sum = 0
while current <= limit:
    if current % 2 == 0:
        print(f'{current} is even')
        even_sum += current
    else:
        print(f'{current} is odd')
        odd_sum += current
    current += 1
print(f'The sum of all even numbers is {even_sum}')
print(f'The sum of all odd numbers is {odd_sum}')
```

```python
# 14. Create a simple calculator using loops (The program should be menu driven which runs till user exits itself.

running = True
while running:
    print('\nCalculator Menu:')
    print('1. Addition')
    print('2. Subtraction')
    print('3. Multiplication')
    print('4. Division')
    print('5. Exit')

    choice = int(input('Enter your choice (1-5): '))

    if choice == 1:
        first_num = int(input('Enter first number: '))
        second_num = int(input('Enter second number: '))
        result = first_num + second_num
        print(f'Result: {result}')
    elif choice == 2:
        first_num = int(input('Enter first number: '))
        second_num = int(input('Enter second number: '))
        result = first_num - second_num
        print(f'Result: {result}')
    elif choice == 3:
        first_num = int(input('Enter first number: '))
        second_num = int(input('Enter second number: '))
        result = first_num * second_num
        print(f'Result: {result}')
    elif choice == 4:
        first_num = int(input('Enter first number: '))
        second_num = int(input('Enter second number: '))
        if second_num == 0:
            print('Division by zero is not allowed')
        else:
            result = first_num / second_num
            print(f'Result: {result}')
    elif choice == 5:
        print('Exiting calculator...')
        running = False
    else:
        print('Invalid choice. Please enter a number between 1 and 5.')
```

```python

# 15. Write a program to check whether the number is automorphic number or not.

number = int(input('Enter a number: '))
square = number * number
temp = number
digit_count = 0
while temp > 0:
    digit_count += 1
    temp //= 10
divisor = 10 ** digit_count
if square % divisor == number:
    print(f'{number} is an automorphic number')
else:
    print(f'{number} is not an automorphic number')
```

```python

# 16. Write a program to reverse a number taken by the user without using strings.

number = int(input('Enter Number: '))
reversed_num = 0
temp = number
while temp > 0:
    reversed_num = reversed_num * 10 + temp % 10
    temp //= 10
print(f'Original Number: {number}')
print(f'Reversed Number: {reversed_num}')
```

```python
 #17. Write a program to count the number of alphabets and no of digits in a string using loops.

text = input('Enter any String: ')
alpha_count = 0
digit_count = 0
position = 0
while position < len(text):
    if text[position].isalpha():
        alpha_count += 1
    elif text[position].isdigit():
        digit_count += 1
    position += 1
print(f'String: {text} and length: {len(text)}')
print(f'Alphabets count: {alpha_count}')
print(f'Digits Count: {digit_count}')
```

```python
# 18. Write a program to check whether the number given by the user is an Armstrong number or not.

number = int(input('Enter a number: '))
original = number
cube_sum = 0
while number > 0:
    digit = number % 10
    cube_sum += digit ** 3
    number //= 10
if original == cube_sum:
    print(f'{original} is an Armstrong number')
else:
    print(f'{original} is not an Armstrong number')


```

```python

# 19. Write a program to check whether the given string is palindrome or not.

text = input('Enter the string: ')
reversed_text = ''
index = 0
while index < len(text):
    reversed_text += text[len(text) - index - 1]
    index += 1
if reversed_text == text:
    print(f'{text} is palindrome')
else:
    print(f'{text} is not palindrome')
```

```python
# 20. Write a program to check whether the given number is palindrome or not.
# Example: 343, 121, 111, etc.
number = int(input('Enter Number: '))
reversed_num = 0
original = number
while number > 0:
    reversed_num = reversed_num * 10 + number % 10
    number //= 10
if reversed_num == original:
    print(f'{original} is palindrome')
else:
    print(f'{original} is not palindrome')
```

```python
# 21. Create Rock, Paper, Scissor game using while loop
# (Hint- use random module and random.choice(['rock', 'paper', 'scissor']) function to get random value selected by computer)
# Your Program will contain one you and second the computer who will be generating the values from its side using random module.
# User will provide the choice of Rock, paper or scissor.
# Print Appropriate Messages and final score of both the computer and user.

import random

print('Game Instructions'.center(50, '-'))
print('You will get 5 chances and scores will be recorded')
print('Enter 0 at any time to exit the game')

player_score = 0
computer_score = 0
round_num = 1

while round_num <= 5:
    print(f'\n--- Round {round_num} ---')
    print('0. Exit Game')
    print('1. Rock')
    print('2. Paper')
    print('3. Scissor')

    player_choice = int(input('Enter your choice: '))

    # Exit condition
    if player_choice == 0:
        print('\nExiting game...')
        break

    comp_selection = random.choice(['rock', 'paper', 'scissor'])

    if player_choice == 1:
        user_pick = 'rock'
    elif player_choice == 2:
        user_pick = 'paper'
    elif player_choice == 3:
        user_pick = 'scissor'
    else:
        print('Invalid choice! Please try again.')
        continue

    print(f'You Selected: {user_pick}')
    print(f'Computer Selected: {comp_selection}')

    if user_pick == comp_selection:
        player_score += 1
        computer_score += 1
        print(f'Tie! Your Score: {player_score} | Computer Score: {computer_score}')
    elif (user_pick == 'rock' and comp_selection == 'scissor') or \
         (user_pick == 'paper' and comp_selection == 'rock') or \
         (user_pick == 'scissor' and comp_selection == 'paper'):
        player_score += 1
        print(f'You Win! Your Score: {player_score} | Computer Score: {computer_score}')
    else:
        computer_score += 1
        print(f'Computer Wins! Your Score: {player_score} | Computer Score: {computer_score}')

    round_num += 1

print('\n' + '=' * 50)
print('Final Score')
print(f'Computer Score: {computer_score}')
print(f'Your Score: {player_score}')
if computer_score > player_score:
    print('Computer Won the Game!')
elif computer_score < player_score:
    print('You Won the Game!')
else:
    print('Match Tie!')
print('Thank you for playing!')
```

```python
# 22. Write a Python program that iterates the integers from 1 to 50. For multiples of three print "Fizz"
# instead of the number and for multiples of five print "Buzz". For numbers that are multiples of three and five,
# print "FizzBuzz".
# Sample Output:
# 1
# 2
# fizz
# 4
# buzz
number = 1
while number <= 50:
    if number % 3 == 0 and number % 5 == 0:
        print('FizzBuzz')
    elif number % 3 == 0:
        print('Fizz')
    elif number % 5 == 0:
        print('Buzz')

    number += 1
```

```python
# 23. Take a string input from user and count the number of vowels and consonants using while loop and for loop both.

sentence = input('Enter a word or a sentence: ')
print(sentence)

i = 0
vowels = 0
consonants = 0

while i < len(sentence):
    if sentence[i].isalpha():  # Check if character is a letter
        if sentence[i].lower() in 'aeiou':
            vowels += 1
        else:
            consonants += 1
    i += 1

print(f'Number of vowels: {vowels}')
print(f'Number of consonants: {consonants}')
```

```python

# 24. Write a program to reverse the String using while loop.
sentence = input('Enter string: ')
i = len(sentence) - 1
mirrored = ''

while i >= 0:
    mirrored += sentence[i]
    i -= 1

print(f'Original string: {sentence}')
print(f'Reversed string: {mirrored}')
```

```python
# 25. Write a Python program to check the validity of passwords input by users.
# Validation:
# - At least 1 letter between [a-z] and 1 letter between [A-Z].
# - At least 1 number between [0-9].
# - At least 1 character from [$#@].
# - Minimum length 6 characters.
# - Maximum length 16 characters.

def checkPassword(password):
    print(password)
    contains_lower = False
    contains_upper = False
    contains_number = False
    contains_special_character = False
    valid_length = False

    if len(password) >= 6 and len(password) <= 16:
        valid_length = True
        for letter in password:
            if letter.islower():
                contains_lower = True
            elif letter.isupper():
                contains_upper = True
            elif letter.isdigit():
                contains_number = True
            elif letter in '$#@':
                contains_special_character = True
    if valid_length and contains_lower and contains_upper and contains_number and contains_special_character:
        print('Valid Password')
    else:
        print('Invalid Password: Please enter a valid password and Try again')

password = input('Enter the password: ') # Rabbit@zoo123 --valid, RabbitAndSnake -- invalid
checkPassword(password)
```
