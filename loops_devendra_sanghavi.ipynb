```python
# 20. Write a program to check whether the given number is palindrome or not.
# Example: 343, 121, 111, etc.
number = int(input('Enter Number: '))
reversed_num = 0
original = number
while number > 0:
    reversed_num = reversed_num * 10 + number % 10
    number //= 10
if reversed_num == original:
    print(f'{original} is palindrome')
else:
    print(f'{original} is not palindrome')
```

```python
# 21. Create Rock, Paper, Scissor game using while loop
# (Hint- use random module and random.choice(['rock', 'paper', 'scissor']) function to get random value selected by computer)
# Your Program will contain one you and second the computer who will be generating the values from its side using random module.
# User will provide the choice of Rock, paper or scissor.
# Print Appropriate Messages and final score of both the computer and user.

import random

print('Game Instructions'.center(50, '-'))
print('You will get 5 chances and scores will be recorded')
print('Enter 0 at any time to exit the game')

player_score = 0
computer_score = 0
round_num = 1

while round_num <= 5:
    print(f'\n--- Round {round_num} ---')
    print('0. Exit Game')
    print('1. Rock')
    print('2. Paper')
    print('3. Scissor')

    player_choice = int(input('Enter your choice: '))

    # Exit condition
    if player_choice == 0:
        print('\nExiting game...')
        break

    comp_selection = random.choice(['rock', 'paper', 'scissor'])

    if player_choice == 1:
        user_pick = 'rock'
    elif player_choice == 2:
        user_pick = 'paper'
    elif player_choice == 3:
        user_pick = 'scissor'
    else:
        print('Invalid choice! Please try again.')
        continue

    print(f'You Selected: {user_pick}')
    print(f'Computer Selected: {comp_selection}')

    if user_pick == comp_selection:
        player_score += 1
        computer_score += 1
        print(f'Tie! Your Score: {player_score} | Computer Score: {computer_score}')
    elif (user_pick == 'rock' and comp_selection == 'scissor') or \
         (user_pick == 'paper' and comp_selection == 'rock') or \
         (user_pick == 'scissor' and comp_selection == 'paper'):
        player_score += 1
        print(f'You Win! Your Score: {player_score} | Computer Score: {computer_score}')
    else:
        computer_score += 1
        print(f'Computer Wins! Your Score: {player_score} | Computer Score: {computer_score}')

    round_num += 1

print('\n' + '=' * 50)
print('Final Score')
print(f'Computer Score: {computer_score}')
print(f'Your Score: {player_score}')
if computer_score > player_score:
    print('Computer Won the Game!')
elif computer_score < player_score:
    print('You Won the Game!')
else:
    print('Match Tie!')
print('Thank you for playing!')
```

```python
# 22. Write a Python program that iterates the integers from 1 to 50. For multiples of three print "Fizz"
# instead of the number and for multiples of five print "Buzz". For numbers that are multiples of three and five,
# print "FizzBuzz".
# Sample Output:
# 1
# 2
# fizz
# 4
# buzz
number = 1
while number <= 50:
    if number % 3 == 0 and number % 5 == 0:
        print('FizzBuzz')
    elif number % 3 == 0:
        print('Fizz')
    elif number % 5 == 0:
        print('Buzz')

    number += 1
```

```python
# 23. Take a string input from user and count the number of vowels and consonants using while loop and for loop both.

sentence = input('Enter a word or a sentence: ')
print(sentence)

i = 0
vowels = 0
consonants = 0

while i < len(sentence):
    if sentence[i].isalpha():  # Check if character is a letter
        if sentence[i].lower() in 'aeiou':
            vowels += 1
        else:
            consonants += 1
    i += 1

print(f'Number of vowels: {vowels}')
print(f'Number of consonants: {consonants}')
```

```python

# 24. Write a program to reverse the String using while loop.
sentence = input('Enter string: ')
i = len(sentence) - 1
mirrored = ''

while i >= 0:
    mirrored += sentence[i]
    i -= 1

print(f'Original string: {sentence}')
print(f'Reversed string: {mirrored}')
```

```python
# 25. Write a Python program to check the validity of passwords input by users.
# Validation:
# - At least 1 letter between [a-z] and 1 letter between [A-Z].
# - At least 1 number between [0-9].
# - At least 1 character from [$#@].
# - Minimum length 6 characters.
# - Maximum length 16 characters.

def checkPassword(password):
    print(password)
    contains_lower = False
    contains_upper = False
    contains_number = False
    contains_special_character = False
    valid_length = False

    if len(password) >= 6 and len(password) <= 16:
        valid_length = True
        for letter in password:
            if letter.islower():
                contains_lower = True
            elif letter.isupper():
                contains_upper = True
            elif letter.isdigit():
                contains_number = True
            elif letter in '$#@':
                contains_special_character = True
    if valid_length and contains_lower and contains_upper and contains_number and contains_special_character:
        print('Valid Password')
    else:
        print('Invalid Password: Please enter a valid password and Try again')

# Call the function AFTER defining it
password = input('Enter the password: ') # Rabbit@zoo123 --valid, RabbitAndSnake -- invalid
checkPassword(password)
```
